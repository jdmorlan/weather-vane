## What a notebook is

A notebook is a **stack of cells**. A cell holds either writing (like this one) or code.

You run a code cell by clicking it and pressing **Shift+Enter**. The answer appears underneath it.

That's the entire interface. Try it — run the cell below.

```python
2 + 3
```

The number that appeared is the value of the *last line* of the cell. You don't need `print`
for that last line. This trips everyone up once and then never again.

Cells remember each other. Run this one, then the one after it:

```python
temperature = 73
```

```python
temperature + 10
```

The second cell knew what `temperature` was because the first one ran. That shared memory is
called the **kernel**. It's a Python session sitting behind the page, holding everything you've
made so far.

> If things ever get confusing, `Kernel → Restart Kernel and Run All Cells` wipes that memory and
> starts clean. It fixes a genuinely surprising number of problems.

## Getting one day of weather

Your station records a reading **every minute**. So one day is about 1,440 readings.

We want 2026-05-10. You'll remember it — 2 inches of rain and 1,516 lightning strikes.

```python
from vane import Archive

arc = Archive()
day = arc.samples('2026-05-10')
```

Nothing printed, because that cell's last line didn't produce a value — it *stored* one, in `day`.

Let's see how big it is. `.shape` gives back (rows, columns).

```python
day.shape
```

---
## What you're holding

`day` is a **DataFrame**. The honest description: *a spreadsheet that lives in a variable.*

Rows, columns, headers. That's it. `.head()` shows the first few rows so you don't print 1,425 of them.

```python
day.head()
```

Read that output slowly. It's worth it.

- The **leftmost column in bold** is the index — here it's the timestamp. Each row is one moment.
- Every other column is a thing the station measured at that moment.
- `NaN` would mean *not measured*. It does **not** mean zero. (Your station has a good day here.)

The column names are a bit cryptic. Here are the ones worth knowing today:

| column | meaning |
|---|---|
| `airTemp` | temperature, °F |
| `windGust` | strongest gust in that minute, mph |
| `precip` | rain that fell in that minute, inches |
| `strikeCount` | lightning strikes in that minute |
| `seaLevelPressure` | air pressure, inHg |

---
## One column at a time

Pull out a single column with square brackets and its name:

```python
day['airTemp']
```

A single column is called a **Series**. A DataFrame is just a bundle of Series that share an index.

That's the whole vocabulary: **Series** = one column. **DataFrame** = a table of them. You now know
the two nouns that 90% of pandas talk uses.

---
## Asking it questions

Now the good part. A Series will answer questions about itself.

Run these one at a time:

```python
day['airTemp'].max()
```

```python
day['airTemp'].min()
```

```python
day['airTemp'].mean()
```

That's the high, the low, and the average for the day — from 1,425 individual readings that you
never had to loop over.

> **Why does it say `np.float32(88.0)`?** It's the number 88.0 with a label saying how it's stored.
> Ignore it — `np.float32(88.0)` and `88.0` are the same number, and it behaves like one everywhere.
> You'll stop seeing it once it's in a chart or a table.

**This is the thing to internalize.** You don't write loops. You say what you want from the whole
column and pandas does the walking. Almost everything that looks intimidating later is this same
move wearing a costume.

Try it yourself — change `airTemp` to `windGust` below and run it:

```python
day['windGust'].max()
```

A few more that answer real questions about that day:

```python
day['precip'].sum()          # total rain, inches
```

```python
day['strikeCount'].sum()     # lightning strikes all day
```

---
## A picture

Any Series can draw itself. One method, no setup:

```python
day['airTemp'].plot()
```

There's your day. Cool at midnight, warming through the morning, and then something happens.

Let's make it readable — a title and a label. Everything after the first line is decoration:

```python
ax = day['airTemp'].plot(figsize=(11, 4), color='#c1121f')
ax.set_title('Temperature, 2026-05-10')
ax.set_ylabel('°F')
```

---
## Now look at the storm

You know a storm hit that day. Plot the rain and see if you can find it.

```python
ax = day['precip'].plot(figsize=(11, 3), color='#4a7ba7')
ax.set_title('Rain per minute, 2026-05-10')
ax.set_ylabel('inches')
```

The flat stretches are dry minutes. The spikes are the storm.

One last one. Plot the temperature and the rain together and look at what the temperature does
*at the moment* the rain starts:

```python
cols = ['airTemp', 'precip']
day[cols].plot(figsize=(11, 5), subplots=True, color=['#c1121f', '#4a7ba7'])
```

There's the story of the day in one picture.

The temperature climbs to 88°F, drifts down through the afternoon — and then falls off a cliff
exactly when the rain arrives. **81°F at 5:15pm, 65°F at 5:45pm.** Sixteen degrees in half an hour.

That's the storm's downdraft: cold air dropping out of the cloud and spreading along the ground.
You felt that happen. It's in your data, and you just found it.

Note what you did there: `day[cols]` with a **list** of names gave you a smaller DataFrame instead
of a Series. One name = one column = Series. A list of names = a table = DataFrame.
